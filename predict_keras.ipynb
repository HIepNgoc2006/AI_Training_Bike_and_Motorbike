# DỰ ĐOÁN THỰC TẾ (INFERENCE)

Script này dùng để nạp lại mô hình đã huấn luyện (`best_keras_model.keras`) và dự đoán thử các bức ảnh thực tế.

## CELL 1: Tải Thư viện và Nạp Mô hình
Chuẩn bị môi trường, sử dụng `%pip` để tương thích `.venv`.

```python
import os
import tensorflow as tf
import numpy as np
import urllib.request
import matplotlib.pyplot as plt

PROJECT_DIR = os.path.abspath(os.getcwd())
MODEL_PATH = os.path.join(PROJECT_DIR, 'best_keras_model.keras')

print("Đang nạp mô hình vào trí nhớ...")
if os.path.exists(MODEL_PATH):
    model = tf.keras.models.load_model(MODEL_PATH)
    print("Nạp mô hình thành công!")
else:
    print(f"LỖI: Không tìm thấy file {MODEL_PATH}. Hãy chạy file train_keras.ipynb trước.")
    model = None
```

## CELL 2: Định nghĩa Hàm Dự Đoán
Hàm này nhận đường dẫn ảnh, tiền xử lý và in kết quả (tỷ lệ % tự tin) ra màn hình.

```python
def predict_image(image_path, model):
    class_names = ['bike', 'motorbike'] # Tên thư mục lúc ta sắp xếp ở train
    
    # Keras tự động load và resize ảnh về 224x224
    img = tf.keras.utils.load_img(image_path, target_size=(224, 224))
    img_array = tf.keras.utils.img_to_array(img)
    img_array = tf.expand_dims(img_array, 0)
    
    predictions = model.predict(img_array)
    score = tf.nn.softmax(predictions[0])
    
    predicted_class = class_names[np.argmax(score)].capitalize()
    confidence = 100 * np.max(score)
    
    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"AI Dự Đoán: {predicted_class}\n(Độ tự tin: {confidence:.2f}%)")
    plt.show()
    
    print(f"Chi tiết nội bộ AI nghĩ: Bike ({100*score[0]:.2f}%) | Motorbike ({100*score[1]:.2f}%)")
```

## CELL 3: Chạy Thực Tế
Tự động tải ảnh trên mạng (nếu chưa có ảnh gốc) và bắt đầu nhận diện.

```python
if model:
    test_image_path = 'test_image.jpg'
    
    # Kịch bản tải ảnh mẫu
    if not os.path.exists(test_image_path):
        print(f"[*] Không tìm thấy ảnh '{test_image_path}'. Hệ thống tự tải ảnh xe máy làm mẫu...")
        sample_url = "https://images.unsplash.com/photo-1558981285-6f0c94958bb6?w=400" 
        try:
            urllib.request.urlretrieve(sample_url, test_image_path)
            print("[*] Đã chuẩn bị xong ảnh mẫu!\n")
        except Exception as e:
            print(f"[!] Lỗi khi tự tải ảnh: {e}")
            
    if os.path.exists(test_image_path):
        print("=== BẮT ĐẦU DỰ ĐOÁN ===")
        predict_image(test_image_path, model)
```
